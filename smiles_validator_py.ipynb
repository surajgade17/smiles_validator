```python
from rdkit import Chem, RDLogger
from rdkit.Chem import Descriptors
RDLogger.DisableLog('rdApp.*')    # Silence RDKit internal warnings
```

```python
def validate_smiles(smiles):
    mol = Chem.MolFromSmiles(smiles)    # Returns None if SMILES is invalid
    return mol is not None    # True if valid molecule, False if None
```

```python
def get_basic_info(smiles):
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return None
    return {
        'SMILES': smiles,
        'Formula': Chem.rdMolDescriptors.CalcMolFormula(mol),
        'Weight': round(Descriptors.MolWt(mol), 2),
        'Atoms': mol.GetNumAtoms(),
    }
```

```python
def validate_list(smiles_list):
    results = []
    for smi in smiles_list:
        is_valid = validate_smiles(smi)
        results.append((smi, is_valid))
    return results
```

```python
# ---- Test Molecules ----
test_smiles = [
    'CCO',
    'CC(=O)Oc1ccccc1C(=O)O',
    'c1ccccc1',
    'INVALID_SMILES',
    'CN1C=NC2=C1C(=O)N(C(=O)N2C)C',
]
```

```python
for smi in test_smiles:
    valid = validate_smiles(smi)
    status = "Valid ✅" if valid else "Invalid ❌"
    print(f"[{status}] {smi}")

print("\nBASIC INFO — Aspirin:")
info = get_basic_info('CC(=O)Oc1ccccc1C(=O)O')
if info:
    for key, value in info.items():
        print(f"  {key}: {value}")

print("\nBATCH RESULTS:")
results = validate_list(test_smiles)
for smi, valid in results:
    status = "Valid ✅" if valid else "Invalid ❌"
    print(f"  {status} | {smi}")
```
